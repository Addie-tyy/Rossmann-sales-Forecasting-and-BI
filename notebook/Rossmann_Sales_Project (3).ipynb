For these stores, the competition was scheduled/opened in August 2015, while our sales data ends on July 31, 2015 hence it shows -1 value.
Our current formula is measuring the difference between the sales date and the recorded competition opening date. Therefore:


`-1` means competition opens approximately one month later

`0` competition opens in the same month

`positive` means competition has already been open

A better feature would be: `CompetitionActive` and we can retain `CompetitionAgeMonths` as an analytical feature.

```python
# 1 for Competition already open
# 0 for Competition not yet open / date unknown
master["CompetitionActive"] = (
    master["CompetitionOpenDate"].notna() &
    (master["Date"] >= master["CompetitionOpenDate"])
).astype(int)
```

3 useful features:
1. `CompetitionDateKnown` : 1	Competition opening date is known, 0	Competition opening date is unknown,
2. `CompetitionActive` : This tells us whether the competition had opened by the sales date. `0` means competition hasn't opened yet, `1`competition has opened, `0*` opening date unknown
3. `CompetitionAgeMonths` : negative values aren't useful, so we can convert negative ages to 0

```python
# Create competition activity flag
master["CompetitionActive"] = (
    (master["CompetitionDateKnown"] == 1) &
    (master["Date"] >= master["CompetitionOpenDate"])
).astype(int)

# Convert negative competition age to 0
master["CompetitionAgeMonths"] = master["CompetitionAgeMonths"].clip(lower=0)

# Check results
master[
    [
        "Store",
        "Date",
        "CompetitionOpenDate",
        "CompetitionDateKnown",
        "CompetitionActive",
        "CompetitionAgeMonths"
    ]
].head(20)
```

# Data Engineering Milestone: Relational Dimensional Modeling (Star Schema)

To prepare for PostgreSQL loading and analytical query optimization, the denormalized dataset was split into an industry-standard Star Schema:
1. `dim_store` (1,115 rows × 11 columns): Store-level dimension capturing metadata, assortment, and competition characteristics.
2. `dim_date` (942 rows × 8 columns): Calendar dimension providing structured temporal slicing (quarter, week of year, weekend flag).
3. `fct_daily_sales` (1,017,209 rows × 11 columns): Core transactional fact table linked via primary and foreign key constraints (`Store`, `Date`).

This structure eliminates data redundancy, enforces relational integrity, and ensures optimal indexing performance in the database.

# Database Schema Preparation: Schema Naming & Type Alignment

Column names across all three Star Schema entities have been standardized into ANSI SQL-compliant lowercase snake_case formatting:
- Keys established: Primary key `store` on `dim_store`, primary key `date` on `dim_date`, and composite foreign key (`store`, `date`) on `fct_daily_sales`.
- Clean separation between dimensions (descriptive metadata) and facts (numerical measures and active state flags).

```python
# create Star Schema tables in PostgreSQL
ddl_schema = """
DROP TABLE IF EXISTS fct_daily_sales CASCADE;
DROP TABLE IF EXISTS dim_store CASCADE;
DROP TABLE IF EXISTS dim_date CASCADE;

-- 1. Dimension Table: Store
CREATE TABLE dim_store (
    store INT PRIMARY KEY,
    store_type VARCHAR(5),
    assortment VARCHAR(5),
    competition_distance DOUBLE PRECISION,
    competition_open_since_month DOUBLE PRECISION,
    competition_open_since_year DOUBLE PRECISION,
    promo2 INT,
    promo2_since_week DOUBLE PRECISION,
    promo2_since_year DOUBLE PRECISION,
    promo_interval VARCHAR(50),
    competition_date_known INT
);

-- 2. Dimension Table: Date
CREATE TABLE dim_date (
    date DATE PRIMARY KEY,
    year INT,
    month INT,
    day INT,
    week_of_year INT,
    quarter INT,
    day_of_week INT,
    is_weekend INT
);

-- 3. Fact Table: Daily Transactions
CREATE TABLE fct_daily_sales (
    store INT REFERENCES dim_store(store),
    date DATE REFERENCES dim_date(date),
    sales INT,
    customers INT,
    open INT,
    promo INT,
    state_holiday VARCHAR(5),
    school_holiday INT,
    competition_active INT,
    competition_age_months DOUBLE PRECISION,
    promo2_active INT,
    PRIMARY KEY (store, date)
);

-- Performance Indexes
CREATE INDEX idx_sales_date ON fct_daily_sales(date);
CREATE INDEX idx_sales_store ON fct_daily_sales(store);
"""

with engine.begin() as conn:
  conn.execute(text(ddl_schema))
  print("All 3 tables and indexes created successfully in PostgreSQL!")
```

# Database Schema Initialization: DDL Execution

Created normalized Star Schema tables in PostgreSQL:
- `dim_store`: Store dimension with primary key constraint on `store`.
- `dim_date`: Calendar dimension with primary key constraint on `date`.
- `fct_daily_sales`: Core transactions with composite primary key `(store, date)`, foreign key relationships linking to both dimensions, and B-tree indexes on `date` and `store`.

```python
import io


def fast_pg_copy(df, table_name, engine):
  """Fastest bulk upload method for PostgreSQL using COPY via in-memory buffer."""
  raw_conn = engine.raw_connection()
  cursor = raw_conn.cursor()
  buffer = io.StringIO()

  # Write DataFrame as CSV format into memory buffer
  df.to_csv(buffer, index=False, header=False, na_rep="\\N")
  buffer.seek(0)

  col_names = ",".join(df.columns.tolist())
  copy_sql = (
      f"COPY {table_name} ({col_names}) FROM STDIN WITH (FORMAT CSV, NULL"
      " '\\N')"
  )

  cursor.copy_expert(copy_sql, buffer)
  raw_conn.commit()
  cursor.close()
  raw_conn.close()
  print(f"Successfully loaded {len(df):,} rows into '{table_name}'")


# 1. Load Dimensions First (Foreign Key dependencies)
fast_pg_copy(dim_store, "dim_store", engine)
fast_pg_copy(dim_date, "dim_date", engine)

# 2. Load Fact Table
fast_pg_copy(fct_daily_sales, "fct_daily_sales", engine)
```

```python
# KPI Query 1: Promotional Uplift & Interaction Analysis (Promo vs Promo2)
kpi_promo_query = """
SELECT
    f.promo AS standard_promo,
    f.promo2_active AS continuous_promo2,
    COUNT(*) AS trading_days,
    ROUND(AVG(f.sales)::numeric, 2) AS avg_sales,
    ROUND(AVG(f.customers)::numeric, 2) AS avg_customers,
    ROUND((AVG(f.sales) / NULLIF(AVG(f.customers), 0))::numeric, 2) AS sales_per_customer
FROM fct_daily_sales f
WHERE f.open = 1
GROUP BY f.promo, f.promo2_active
ORDER BY avg_sales DESC;
"""

with engine.connect() as conn:
  result = conn.execute(text(kpi_promo_query))
  promo_kpi_df = pd.DataFrame(result.fetchall(), columns=result.keys())

promo_kpi_df
```

```python
# KPI Query 2: Competition Proximity Impact on Store Sales & Footfall
kpi_competition_query = """
SELECT
    CASE
        WHEN s.competition_distance < 1000 THEN '1. Very Close (<1km)'
        WHEN s.competition_distance BETWEEN 1000 AND 2999 THEN '2. Moderate (1-3km)'
        WHEN s.competition_distance BETWEEN 3000 AND 9999 THEN '3. Distant (3-10km)'
        ELSE '4. Very Distant (>10km)'
    END AS distance_tier,
    COUNT(DISTINCT s.store) AS total_stores,
    COUNT(*) AS total_trading_days,
    ROUND(AVG(f.sales)::numeric, 2) AS avg_sales,
    ROUND(AVG(f.customers)::numeric, 2) AS avg_customers,
    ROUND((AVG(f.sales) / NULLIF(AVG(f.customers), 0))::numeric, 2) AS sales_per_customer
FROM fct_daily_sales f
JOIN dim_store s ON f.store = s.store
WHERE f.open = 1
GROUP BY 1
ORDER BY 1;
"""

with engine.connect() as conn:
  result = conn.execute(text(kpi_competition_query))
  comp_kpi_df = pd.DataFrame(result.fetchall(), columns=result.keys())

comp_kpi_df
```

3. **Basket Size vs. Footfall Trade-Off**: Customer basket value scales inversely with competition distance: from **7.90 sales/customer** in dense zones (<1km) to **10.34 sales/customer** in isolated zones (>10km), indicating frequent convenience visits downtown versus planned, high-basket trip behavior in suburban/rural stores.

```python
# KPI Query 3: Time-Series Window Analytics (7-Day Moving Avg & Sales Velocity)
kpi_timeseries_query = """
WITH daily_store_metrics AS (
    SELECT
        f.store,
        f.date,
        d.day_of_week,
        d.is_weekend,
        f.open,
        f.sales,
        f.customers,
        -- 7-Day Moving Average (including closed days to track true weekly velocity)
        ROUND(AVG(f.sales) OVER (
            PARTITION BY f.store
            ORDER BY f.date
            ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
        )::numeric, 2) AS sales_7d_mavg,

        -- Prior Day Sales (Lag)
        LAG(f.sales, 1) OVER (
            PARTITION BY f.store
            ORDER BY f.date
        ) AS prior_day_sales
    FROM fct_daily_sales f
    JOIN dim_date d ON f.date = d.date
)
SELECT
    store,
    date,
    day_of_week,
    is_weekend,
    open,
    sales,
    prior_day_sales,
    (sales - prior_day_sales) AS dod_sales_change,
    sales_7d_mavg
FROM daily_store_metrics
WHERE store = 1
ORDER BY date DESC
LIMIT 15;
"""

with engine.connect() as conn:
  result = conn.execute(text(kpi_timeseries_query))
  timeseries_df = pd.DataFrame(result.fetchall(), columns=result.keys())

timeseries_df
```

# SQL Analytics Finding 3: Time-Series Seasonality & Moving Average Smoothing

1. **Monday Rebound Velocity**: Sunday closures (`open = 0`) induce a predictable demand spike on Mondays (`day_of_week = 1`), generating peak weekday sales (~6,102 vs ~3,500–5,000 mid-week) and a massive Day-over-Day delta.
2. **Window Function Utility**: A 7-day rolling window (`ROWS BETWEEN 6 PRECEDING AND CURRENT ROW`) effectively filters out periodic weekend closures, exposing the underlying sales trend (Store 1 volume expanded from ~3,322 to ~4,363 in late July 2015).
3. **Forecasting Pre-condition**: Because Sunday closure is mandatory for most standard stores, models must explicitly condition lag features on operational open states rather than raw chronological lags.

```python
# Extract Clean Modeling Dataset from PostgreSQL (Excluding Closed Stores)
ml_extract_query = """
SELECT
    f.store,
    f.date,
    f.sales,
    f.customers,
    f.promo,
    f.state_holiday,
    f.school_holiday,
    f.competition_active,
    f.competition_age_months,
    f.promo2_active,
    d.year,
    d.month,
    d.day,
    d.week_of_year,
    d.quarter,
    d.day_of_week,
    d.is_weekend,
    s.store_type,
    s.assortment,
    s.competition_distance,
    s.promo2
FROM fct_daily_sales f
JOIN dim_date d ON f.date = d.date
JOIN dim_store s ON f.store = s.store
WHERE f.open = 1 AND f.sales > 0
ORDER BY f.date ASC;
"""

with engine.connect() as conn:
  ml_data = pd.read_sql(text(ml_extract_query), conn)

print("ML Dataset Shape:", ml_data.shape)
print("Date Range:", ml_data["date"].min(), "to", ml_data["date"].max())
ml_data.head()
```

# Modeling Dataset Preparation: Active Trade Filtering & Chronological Framing

1. **Active Trading Cohort**: Filtering for `open = 1` and `sales > 0` produces **844,338 valid training instances**, removing closed-day structural zeros and rare anomalies.
2. **Holiday Operation Signals**: Early dates like `2013-01-01` illustrate that `StoreType b` predominantly captures trading volume on public holidays while formats `a`, `c`, and `d` remain shut.
3. **Validation Strategy**: Because retail sales contain strong trend and seasonality components, traditional random k-fold splitting would cause lookahead leakage. A time-based out-of-time validation split (holding out the final 6 weeks of data, June 19 to July 31, 2015) will be implemented to replicate the test evaluation period.

```python
# Feature Preprocessing and Chronological Train/Validation Split

# 1. Handle missing values for model compatibility
ml_data["competition_age_months"] = ml_data["competition_age_months"].fillna(0)
ml_data["state_holiday"] = ml_data["state_holiday"].astype(str)

# 2. Encode categorical columns
categorical_cols = ["store_type", "assortment", "state_holiday"]
for col in categorical_cols:
  ml_data[col] = ml_data[col].astype("category")

# 3. Define the 6-week out-of-time validation cutoff (mirroring the test set)
val_cutoff = pd.to_datetime("2015-06-19").date()
ml_data["date"] = pd.to_datetime(ml_data["date"]).dt.date

train_mask = ml_data["date"] < val_cutoff
val_mask = ml_data["date"] >= val_cutoff

print(f"Train split size: {train_mask.sum():,} records")
print(f"Validation split size: {val_mask.sum():,} records")
print(
    f"Validation date range: {ml_data.loc[val_mask, 'date'].min()} to"
    f" {ml_data.loc[val_mask, 'date'].max()}"
)
```

# Modeling Strategy: Chronological Split & RMSPE Metric Formulation

1. **Test Set Alignment**: An out-of-time 6-week validation set (June 19, 2015 – July 31, 2015; 41,396 records) replicates the exact forecast length of the deployment test set.
2. **Evaluation Metric**: Predictions are evaluated using Root Mean Squared Percentage Error (RMSPE), penalizing relative errors uniformly across both high-volume flagships and low-volume local stores.
3. **Target Transformation**: The model trains on `log1p(Sales)` to normalize target skewness and approximate RMSPE minimization via standard MSE objective functions.

```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor

# 1. Feature selection (excluding customers because it's unknown in future test periods)
features = [
    "store",
    "day_of_week",
    "promo",
    "state_holiday",
    "school_holiday",
    "competition_active",
    "competition_age_months",
    "promo2_active",
    "year",
    "month",
    "day",
    "week_of_year",
    "quarter",
    "is_weekend",
    "store_type",
    "assortment",
    "competition_distance",
    "promo2",
]

target = "sales"

# 2. Convert category types to integer codes for scikit-learn compatibility
X_train = ml_data.loc[train_mask, features].copy()
X_val = ml_data.loc[val_mask, features].copy()

categorical_cols = ["store_type", "assortment", "state_holiday"]
for col in categorical_cols:
  X_train[col] = X_train[col].cat.codes
  X_val[col] = X_val[col].cat.codes

# Train on log1p(sales) to mirror percentage error optimization
y_train = np.log1p(ml_data.loc[train_mask, target])
y_val = ml_data.loc[val_mask, target].values  # True raw sales for RMSPE


# 3. RMSPE evaluation function
def rmspe(y_true, y_pred):
  pct_error = (y_true - y_pred) / y_true
  return np.sqrt(np.mean(np.square(pct_error)))


# 4. Initialize and fit Random Forest
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=5,
    max_samples=0.5,
    random_state=42,
    n_jobs=-1,
    verbose=1,
)

print("Training Random Forest...")
rf_model.fit(X_train, y_train)

# 5. Predict and evaluate
val_preds_rf = np.expm1(rf_model.predict(X_val))
val_rmspe_rf = rmspe(y_val, val_preds_rf)

print(
    f"\nRandom Forest Validation RMSPE: {val_rmspe_rf:.4f}"
    f" ({val_rmspe_rf*100:.2f}%)"
)
```

# Machine Learning Forecasting: Random Forest Benchmark & Evaluation

1. **Validation Performance**: The Random Forest regressor achieves an out-of-time **RMSPE of 0.1648 (16.48%)** across the 6-week holdout period (June 19 – July 31, 2015).
2. **Generalization Integrity**: Because the test split is purely chronological, this 16.48% error rate demonstrates robust cross-store generalization without forward-looking data leakage.
3. **Execution Profile**: Leveraging sub-sampling (`max_samples=0.5`) and depth pruning (`max_depth=20`, `min_samples_leaf=5`) allowed fitting 100 estimators over 802,942 records in 3.0 minutes.
